# 08. Đánh giá DistilBERT bằng nhiều lần chia hoặc K-Fold

Mỗi lần chia/fold khởi tạo một mô hình mới từ `distilbert-base-uncased`. Kết quả được tổng hợp bằng trung bình và độ lệch chuẩn. Tập test của một split không được dùng để huấn luyện trong chính split đó.

```
from pathlib import Path
import sys

PROJECT_ROOT = Path('..').resolve()
SRC_DIR = PROJECT_ROOT / 'src'
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

from common import set_global_seed

SEED = 42
set_global_seed(SEED)
print(f"Đã thiết lập seed tái lập: {SEED}")
```

## Repeated Stratified Holdout

Ba seed 42, 52 và 62 là cấu hình khuyến nghị khi tài nguyên CPU hạn chế.

```
RUN_REPEATED_HOLDOUT = False

if RUN_REPEATED_HOLDOUT:
    import subprocess
    subprocess.run(
        [
            sys.executable,
            'src/evaluate_distilbert.py',
            '--protocol', 'repeated_holdout',
            '--seeds', '42', '52', '62',
        ],
        cwd=PROJECT_ROOT,
        check=True,
    )
else:
    print('Đặt RUN_REPEATED_HOLDOUT=True để chạy 3 lần chia DistilBERT.')
```

## Stratified K-Fold

3-Fold phù hợp hơn 5-Fold khi chỉ có CPU. Không cần chạy đồng thời cả hai giao thức trong báo cáo; chọn một và ghi rõ.

```
RUN_KFOLD = False

if RUN_KFOLD:
    import subprocess
    subprocess.run(
        [
            sys.executable,
            'src/evaluate_distilbert.py',
            '--protocol', 'kfold',
            '--n-splits', '3',
            '--seeds', '42',
        ],
        cwd=PROJECT_ROOT,
        check=True,
    )
else:
    print('Đặt RUN_KFOLD=True để chạy Stratified 3-Fold DistilBERT.')
```
